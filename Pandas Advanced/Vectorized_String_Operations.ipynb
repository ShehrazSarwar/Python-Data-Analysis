```python
# problem in vectorized opertions in vanilla python
s = ['cat','mat',None,'rat']

[i.startswith('c') for i in s]
```

```python
df = pd.read_csv('/content/titanic.csv')
df['Name']
```

```python
# Some String Operations
df['Name'].str.capitalize()
df['Name'].str.title()
df['Name'].str.lower()
df['Name'].str.upper()
```

```python
# len
df['Name'].str.len() # Length of Each String
```

```python
# find largest string
df['Name'].str.len().max()
df['Name'][df['Name'].str.len() == df['Name'].str.len().max()].values[0]
```

```python
# 2nd Method with idxmax()
df['Name'].loc[df['Name'].str.len().idxmax()]
```

```python
# strip
df['Name'].str.strip()
```

```python
# split
df['Surname'] = df['Name'].str.split(',').str.get(0)
```

```python
df[['Title','Firstname']] = df['Name'].str.split(',').str.get(1).str.strip().str.split(n=1,expand=True)
```

```python
df.head()
```

```python
df['Title'].value_counts()
```

```python
# replace
df['Title'] = df['Title'].str.replace('Ms.','Miss.')
df['Title'] = df['Title'].str.replace('Mlle.','Miss.')
```

```python
df['Title'].value_counts()
```

```python
#filtering
# startswith()/endswith()
# isdigit()/isalpha()
df[df['Firstname'].str.startswith('A')]
# Similarly we can check if there is any digits etc
# Same python String methods
```

```python
# contains
# search john boht case
df[df['Firstname'].str.contains('john',case=False)]
```

```python
# apply regex
# find Surnames starts with vowel and ends with vowel
df[df['Surname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
```

```python
# slicing
df['Name'].str[::-1]
```
