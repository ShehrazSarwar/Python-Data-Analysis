```python
import pandas as pd
import numpy as np
```

```python
courses = pd.read_csv('/content/courses.csv')
students = pd.read_csv('/content/students.csv')
nov = pd.read_csv('/content/reg-month1.csv')
dec = pd.read_csv('/content/reg-month2.csv')

matches = pd.read_csv('/content/matches.csv')
delivery = pd.read_csv('/content/deliveries.csv')
```

```python
# pd.concat
# ignore_index
# mullitindex -> fetch using loc
# concat dataframes horizontally
```

```python
regs = pd.concat([nov,dec],ignore_index = True)
regs
```

```python
# Multi-index DataFrame
multi = pd.concat([nov,dec],keys = ['Nov','Dec'])
print(multi.loc[('Nov',4)])
```

```python
# Concat dataframes horizontally
pd.concat([nov,dec],axis = 1)
```

merge() --> inner, outer, left and right

```python
regs
```

```python
# inner join
students.merge(regs,how='inner',on='student_id')
```

```python
regs
```

```python
# left join
courses.merge(regs,how='left',on='course_id')
```

```python
# right join
temp_df = pd.DataFrame({
    'student_id':[26,27,28],
    'name': ['Nitesh','Ali','Jay'],
    'partner':[28,26,27]
})

students = pd.concat([students,temp_df],ignore_index=True)
```

```python
students.merge(regs,how='right',on='student_id')
```

```python
students.merge(regs,how='left',on='student_id')
```

```python
# outer join
students.merge(regs,how='outer',on='student_id').tail(10)
```

```python
# find the total revenue generated
total = regs.merge(courses,how = 'inner',on='course_id')['price'].sum()
print(total)
```

```python
# find the month by month revenue
temp_df = pd.concat([nov,dec],keys=['Nov','Dec']).reset_index()
temp_df.merge(courses,on='course_id').groupby('level_0')['price'].sum() # by default how ='inner'
```

```python
# print the registration table
# cols --> name --> course --> price
regs.merge(students,on='student_id').merge(courses,on='course_id')[['name','course_name','price']]
```

```python
# plot bar chart for revenue/course
temp_df = regs.merge(courses,on='course_id').groupby('course_name')['price'].sum()
temp_df.plot(kind='bar',ylabel='Revenues',xlabel='Courses',rot=45)
```

```python
# find students who enrolled in both the months
# Method 1: return unique values
dec[dec['student_id'].isin(nov['student_id'])]['student_id'].drop_duplicates().sort_values().tolist()
```

```python
# Method 2: return unique values only which present in both months directly
common_student_id = np.intersect1d(nov['student_id'], dec['student_id'])
common_student_id
```

```python
# Method 3
dec.merge(nov,on='student_id')['student_id'].drop_duplicates().sort_values().tolist()
```

```python
students[students['student_id'].isin(common_student_id)]
```

```python
# find course that got no enrollement
# Method 1
temp_df = regs.merge(courses,how = 'right',on='course_id')
temp_df[temp_df['student_id'].isnull()]['course_name']
```

```python
# Method 2
courses_list = np.setdiff1d(courses['course_id'],regs['course_id'])
courses[courses['course_id'].isin(courses_list)]
```

```python
# find students who did not enrolled into any of the courses
# Method 1:
temp_df = students.merge(regs,how ='left',on='student_id')
temp_df[temp_df['course_id'].isnull()][['student_id','name']]
```

```python
# Method 2:
students_list = np.setdiff1d(students['student_id'],regs['student_id'])
students[students['student_id'].isin(students_list)]
```

```python
# Print student name -> partner name for all enrolled students
# self join
students.merge(students,how='inner',left_on= 'partner',right_on='student_id')[['name_x','name_y']]
```

```python
# find top 3 students who did most number of enrollments
students.merge(regs,on='student_id').groupby(['student_id','name'])['name'].count().sort_values(ascending=False).head(3)
```

```python
# Method 2:
students.merge(regs,on='student_id').groupby(['student_id','name']).size().sort_values(ascending=False).reset_index().rename(columns = {0:'enrollments'}).head(3)
```

```python
# Method 3: incase we want all top 3 even if there are ties
students.merge(regs,on='student_id').groupby(['student_id','name']).size().nlargest(3,keep='all')
```

```python
# find top 3 students who spent most amount of money on courses
students.merge(regs,on='student_id').merge(courses,on='course_id').groupby(['student_id','name'])['price'].sum().nlargest(3)
```

```python
# Alternate Syntax For Merge()
# pd.merge()
pd.merge(students,regs,how='inner',on='student_id')
```

```python
temp_df = delivery.merge(matches,left_on='match_id',right_on='id')
```

```python
six_df = temp_df[temp_df['batsman_runs'] == 6]
```

```python
# stadium --> sixes
num_sixes = six_df.groupby('venue')['venue'].count()
```

```python
num_matches = matches['venue'].value_counts()
```

```python
(num_sixes/num_matches).sort_values(ascending=False).head(3)
```

```python
# Find Orange cap holder of all the seasons
new_df = delivery.merge(matches,left_on='match_id',right_on='id')
new_df.groupby(['season','batsman'])['batsman_runs'].sum().reset_index().sort_values('batsman_runs',ascending=False).drop_duplicates(subset='season').sort_values('season')
```
