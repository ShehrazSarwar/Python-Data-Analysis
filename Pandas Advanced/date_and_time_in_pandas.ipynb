### Timestamp Object

Time stamps reference particular moments in time (e.g., Oct 24th, 2022 at 7:00pm)*italicized text*

Different ways to create Timestamp Object

```python
# With /
pd.Timestamp("2025/8/8")
```

```python
# With -
pd.Timestamp("2025-8-8")
```

```python
# With Time and PM
pd.Timestamp("2025-8-8 9:33PM")
```

```python
# Default AM
pd.Timestamp("2025-8-8 9:33")
```

```python
# Text Form
pd.Timestamp("27th january 2005")
pd.Timestamp("27th january 2005 6:39PM")
```

```python
# With Comas Seperated
pd.Timestamp('2023, 1, 5,')
```

```python
# only year
pd.Timestamp('2025')
```

```python
# Using python's DateTime Object
import datetime as dt
```

```python
dt.datetime(2025,8,8,6,44)
```

```python
x = pd.Timestamp(dt.datetime(2025,8,8,6,44))
```

```python
# Attributes of Timestamp
x.year
x.month
x.day
x.hour
x.minute
x.second
```

```python
# Others
x.days_in_month
x.is_leap_year
x.is_month_end
x.is_month_start
x.is_year_end
x.is_quarter_start
x.days_in_month
```

### DatetimeIndex Object

A collection of pandas timestamp

```python
# Using list of Strings
pd.DatetimeIndex(['2025/8/8','2024/1/5','2025/3/3'])
```

```python
# Using List of datetime object of python
pd.DatetimeIndex([dt.datetime(2025,8,8),dt.datetime(2024,1,5),dt.datetime(2025,3,3)])
```

```python
# Using Timestamp objects
dt_index = pd.DatetimeIndex([pd.Timestamp('2025-8-8'),pd.Timestamp('2024-1-5'),pd.Timestamp('2025, 3, 3')])
```

```python
# Using DatetimeIndex as series index
pd.Series([1,2,3], index= dt_index)
```

### date_range() Function

```python
# Generating Dates
pd.date_range(start='2025-1-1', end='2025-2-28')
```

```python
# Alternate Days
pd.date_range(start='2025-1-1', end='2025-2-28',freq='2D') # Similarly 3D For 2 Days Gap
```

```python
# Business Days (Mon-Fri)
pd.date_range(start='2025-1-1', end='2025-2-28',freq='B')
```

```python
# Only First Day Of The Week --> W
pd.date_range(start='2025-1-1', end='2025-2-28',freq='W')
```

```python
pd.date_range(start='2025-1-1', end='2025-2-28',freq='W-MON') # Only Mondays
```

```python
# Only First Day Of The Alternate Weeks --> 2W
pd.date_range(start='2025-1-1', end='2025-2-28',freq='2W')
```

```python
# Hourly Data ---> h
pd.date_range(start='2025-1-1', end='2025-2-28',freq='h')
```

```python
# Month End -- > ME
pd.date_range(start='2025-1-1', end='2025-2-28',freq='ME')
```

```python
# Month End -- > MS
pd.date_range(start='2025-1-1', end='2025-2-28',freq='MS')
```

```python
# YE -- > Year Ends
pd.date_range(start='2023',end='2030',freq='YE')
```

```python
# Using periods --> no of days forward
pd.date_range(start='2025',periods=25)
```

### to_datetime function

converts an existing objects to pandas timestamp/datetimeindex object

```python
# simple Series example
temp_df = pd.Series(['2025/8/8','2024/2/27','2025/4/25'])
pd.to_datetime(temp_df)
```

```python
pd.to_datetime(temp_df).dt.year
pd.to_datetime(temp_df).dt.month_name()
pd.to_datetime(temp_df).dt.day_name()
```

```python
# With Errors
temp_df = pd.Series(['2025/8/8','2024/2/27','2025/44/25'])
pd.to_datetime(temp_df, errors='coerce')
```

```python
pd.to_datetime(temp_df, errors='coerce').dt.month_name()
```
