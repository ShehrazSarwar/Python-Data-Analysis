```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

### Colored Scatter Plot

```python
iris = pd.read_csv('/content/drive/MyDrive/Python_Adv/Data Analysis/Matplotlib/datasets-session-24/iris.csv')
iris.sample(5)
```

```python
iris['Species'].unique()
```

```python
iris.replace({'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2}, inplace = True)
# cmap to change color style and alpha for opactiy (0-1 range) 1 is default
# c for colors
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c= iris['Species'],cmap= 'plasma',alpha= 1)
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.title('Colored Plot Of Sepal and Petal Length')
plt.colorbar()
plt.show()
```

### Annotations

```python
batter = pd.read_csv('/content/drive/MyDrive/Python_Adv/Data Analysis/Matplotlib/datasets-session-24/batter.csv')
df = batter.head(100).sample(25,random_state=5).reset_index(drop=True)
df
```

```python
plt.figure(figsize=(15,10))
plt.scatter(df['avg'],df['strike_rate'],s=df['runs'])
for i in range(len(df)):
  plt.text(df['avg'].iloc[i],df['strike_rate'].iloc[i],s=df['batter'].iloc[i],fontdict={'color':'brown','size':9})

plt.title('Avg VS Strike Rate')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
plt.show()
```

### Horizontal & Vertical Lines

```python
plt.figure(figsize=(15,10))
plt.scatter(df['avg'],df['strike_rate'],s=df['runs'])
for i in range(len(df)):
  plt.text(df['avg'].iloc[i],df['strike_rate'].iloc[i],s=df['batter'].iloc[i],fontdict={'color':'brown','size':9})

plt.axhline(130,color='red')
plt.axvline(30,color='orange')
plt.title('Avg VS Strike Rate')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
plt.show()
```

```python
fig, ax = plt.subplots()
x = np.arange(len(sample_df))
ax.bar(x - 0.2,sample_df['2015'],label = '2015',width = 0.2)
ax.bar(x ,sample_df['2016'],label = '2016',width = 0.2,color='orange')
ax.bar(x + 0.2,sample_df['2017'],label = '2017',width = 0.2,color='red')
ax.set_xticks(x)
ax.set_xticklabels(sample_df['batsman'])
ax.set_title('batsman season record')
ax.set_ylabel('Runs')
ax.set_xlabel('Batsman')
plt.legend()
fig.show()
```

```python
fig, ax = plt.subplots(2,1,sharex=True,figsize=(8,5))
ax[0].scatter(batter['avg'],batter['strike_rate'],color='red',marker='+')
ax[0].set_ylabel('Strike Rate')
ax[0].set_title('Avg Vs Strike Rate')

ax[1].scatter(batter['avg'],batter['runs'])
ax[1].set_ylabel('Runs')
ax[1].set_title('Avg Vs Runs')
ax[1].set_xlabel('Avg')
```

```python
fig, ax = plt.subplots(2,2,figsize=(8,8))
ax[0,0].scatter(batter['avg'],batter['strike_rate'],color='red',marker='+')
ax[1,0].hist(batter['avg'],color='purple')
ax[0,1].scatter(batter['avg'],batter['runs'])
ax[1,1].bar(batter['batter'].iloc[:5],batter['runs'].iloc[:5],color=['orange','red','blue','green','purple'])
ax[1,1].set_xticks(batter['batter'].iloc[:5])
ax[1,1].set_xticklabels(batter['batter'].iloc[:5],rotation=40)
fig.show()
```

### 3D Line Plot

```python
x = [10,20,30,40]
y = [6,12.5,55,9]
z = [50,78,66,44]
fig = plt.figure()
ax = plt.subplot(projection = '3d')
ax.scatter3D(x,y,z,s=[100,100,100,100],color='yellow')
ax.plot3D(x,y,z,color='red')
fig.show()
```

### 3D Surface Plot

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx,yy = np.meshgrid(x,y)
z = xx**2 + yy**2
fig = plt.figure()
ax = plt.subplot(projection = '3d')
p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
fig.show()
```

### Contour Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx,yy = np.meshgrid(x,y)
z = xx**2 + yy**2

fig = plt.figure()
ax = plt.subplot()
p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
fig.show()
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx,yy = np.meshgrid(x,y)
z = xx**2 + yy**2

fig = plt.figure()
ax = plt.subplot()
p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
fig.show()
```
