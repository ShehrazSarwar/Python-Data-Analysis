### **Problem-1:** Make a subplots which have 2 plots.

- For the first chart, draw a scatter plot "*Monitored Cap.(MW)*" vs "*Total Cap. Under Maintenance (MW)*" of top 5 most frequent power stations. Then draw the lines which indicate the average values of these two columns. Change the colors according to the names of the Power Stations.
- For the second chart, draw a scatter plot "*Monitored Cap. (MW)*" vs "*Actual(MU)*" of the top 5 most frequent power stations. Also draw the lines which indicates the average values of these two columns. Change the colors according to the names of the Power Stations.


````python
# code here
df = pd.read_csv('https://rb.gy/gsmddu')
df
````

````python
top_5 = df['Power Station'].value_counts().head().index.values
temp_df = df[df['Power Station'].isin(top_5)]

color_map = {station: i for i, station in enumerate(top_5)}

fig, ax = plt.subplots(2,1,sharex=True,figsize=(12,10))

ax1 = ax[0].scatter(temp_df['Monitored Cap.(MW)'],
                    temp_df['Total Cap. Under Maintenace (MW)'],
                    c=temp_df['Power Station'].map(color_map),
                    cmap='plasma')
ax[0].axhline(temp_df['Total Cap. Under Maintenace (MW)'].mean(),color='Red',linestyle='--')
ax[0].axvline(temp_df['Monitored Cap.(MW)'].mean(),color='Red',linestyle='--')
ax[0].set_ylabel('Total Cap Under Maintenace (MW)')
ax[0].set_title('Monitored Cap.(MW)" vs "Total Cap. Under Maintenance (MW)')
cbar1 = plt.colorbar(ax1)
cbar1.set_ticks(list(color_map.values()))
cbar1.set_ticklabels(list(color_map.keys()))

ax2 = ax[1].scatter(temp_df['Monitored Cap.(MW)'],
                    temp_df['Actual(MU)'],
                    c=temp_df['Power Station'].map(color_map),
                    cmap='plasma')
ax[1].axhline(temp_df['Actual(MU)'].mean(),color='Red',linestyle='--')
ax[1].axvline(temp_df['Monitored Cap.(MW)'].mean(),color='Red',linestyle='--')
ax[1].set_title('Monitored Cap. (MW)" vs "Actual(MU)')
ax[1].set_ylabel('Actual(MU)')
ax[1].set_xlabel('Monitored Cap.(MW)')
cbar2 = plt.colorbar(ax2)
cbar2.set_ticks(list(color_map.values()))
cbar2.set_ticklabels(list(color_map.keys()))

plt.tight_layout()
plt.show()
````

### **Problem-3:** Make a 3D *Surface* plot of this below mathematical equation.

$$z = |x| - |y|$$

````python
# code here
x = np.linspace(-100, 100, 200)
y = np.linspace(-100, 100, 200)
xx, yy = np.meshgrid(x,y)
z = np.abs(xx) - np.abs(yy)
fig = plt.figure(figsize=(12,8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
ax.set_title("3D Surface Plot of z = |x| - |y|")
plt.show()
````

### `Problem-6` Use Pandas plot functions

* Line plot of closing value of top 5 Stocks in Year 2020.
* Take top 5 stocks based on total turnover in Year 2020

````python
# code here
df = pd.read_csv('/content/nifty-50 - nifty-50.csv')
df.head()
````

````python
df['Date'] = pd.to_datetime(df['Date'])
df = df[df['Date'].dt.year == 2020]
````

### `Problem 9:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = sin(\sqrt{x^2 + y^2})$$


````python
# code here
x = np.linspace(0, 100, 200)
y = np.linspace(0, 100, 200)
xx, yy = np.meshgrid(x,y)
z = np.sin(np.sqrt((xx**2 + yy**2)))

fig = plt.figure(figsize=(12,8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
ax.set_title("Surface Plot of z=sin(√x2+y2)")
plt.show()
````

````python
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(111)
ax_obj = ax.contour(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_title("Countour Plot of z=sin(√x2+y2)")
plt.colorbar(ax_obj,label='Z values')
plt.show()
````

````python
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(111)
ax_obj = ax.contourf(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_title("Countourf Plot of z=sin(√x2+y2)")
plt.colorbar(ax_obj,label='Z values')
plt.show()
````

### `Problem 10:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = tan(\log_2({x^2 + y^2})$$

````python
# code here
x = np.linspace(1, 100, 200)
y = np.linspace(1, 100, 200)
xx, yy = np.meshgrid(x,y)
z = np.tan(np.log2(xx**2 + yy**2))

# Mask extreme values for clearer visualization
z = np.where(np.abs(z) > 10, np.nan, z)

fig = plt.figure(figsize=(12,8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")
ax.set_title("Surface Plot of z=tan(log₂(x² + y²))")
plt.show()
````

````python
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(111)
ax_obj = ax.contour(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_title("Countour Plot of z=tan(log₂(x² + y²))")
plt.colorbar(ax_obj,label='Z values')
plt.show()
````

````python
fig = plt.figure(figsize=(10,6))
ax = fig.add_subplot(111)
ax_obj = ax.contourf(xx,yy,z,cmap='viridis')
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_title("Countourf Plot of z=tan(log₂(x² + y²))")
plt.colorbar(ax_obj,label='Z values')
plt.show()
````
