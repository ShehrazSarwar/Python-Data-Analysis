### Types of Data

- Numerical Data
- Categorical Data


Univariate Analysis → Examining one column (feature/variable) at a time.
Example: Distribution of Age in the Titanic dataset.

Bivariate Analysis → Examining the relationship between two variables.
Example: Relationship between Age and Survived.

Multivariate Analysis → Examining more than two variables simultaneously.
Example: Relationship between Age, Fare, and Survived.

```python
# import the library
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

### 2D Line plot

image.png

- Bivariate Analysis
- Categorical and Numercial Data or Both Numerical
- Use Case -> Time Series Data

```python
# Plotting a Simple Function
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)
plt.show()
```

```python
batsman = pd.read_csv('/content/sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.show()
```

```python
# Multiple Plots at Once
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.show()
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.title('Rohit Sharma Vs Virat Kholi Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# colors(hex) and line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='#F57414')
# can provide color name also like purple
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple')
plt.title('Rohit Sharma Vs Virat Kholi Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# linestyle = dotted, dashdot, dashed, solid(default), and etc
# linewidth
plt.plot(batsman['index'],batsman['V Kohli'],color='#F57414',linestyle='solid',linewidth = 3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple',linestyle='dashdot',linewidth = 2)
plt.title('Rohit Sharma Vs Virat Kholi Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# marker and markersize
# marker --> D, +, ., o, etc
# legend --> location
plt.plot(batsman['index'],batsman['V Kohli'],color='#F57414',linestyle='solid',linewidth = 3,marker='D',markersize = 8,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple',linestyle='dashdot',linewidth = 2,marker='o',markersize = 8,label='Rohit')
plt.title('Rohit Sharma Vs Virat Kholi Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend(loc='lower right') # loc = 'best' by default
plt.show()
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,75000)
plt.xlim(2017,2019)
plt.show()
```

```python
# grid
plt.plot(batsman['index'],batsman['V Kohli'],color='#F57414',linestyle='solid',linewidth = 3,marker='D',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='purple',linestyle='dashdot',linewidth = 2,marker='o',label='Rohit')
plt.title('Rohit Sharma Vs Virat Kholi Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.legend()
plt.grid()
plt.show()
```

## Scatter Plot

image.png

- Bivariate Analysis
- numerical vs numerical
- Use case - Finding correlation

```python
x = np.linspace(-10,10,50)
y = 10*x + 3 + np.random.randint(0,300,50)

plt.scatter(x,y)
plt.show()
```

```python
df = pd.read_csv('/content/batter.csv')
df = df.head(50)
df.head()
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
plt.title('Avg and SR Comparision of Top 50 Batsman')
plt.xlabel('Avg Runs')
plt.ylabel('Strike Rate')
plt.show()
```

```python
# size (s)
tips = sns.load_dataset('tips')
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
plt.show()
```

```python
# using plt.plot()
# Advantage: faster for large dataset
# Disadvantage: not include size parameter
plt.plot(tips['total_bill'],tips['tip'],'o')
plt.show()
```

## Bar Chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,8,6,12,7]
colors = ['Red','Blue','Green','Yellow','Pink']

plt.bar(colors,children,color=colors)
plt.title('favourite Colour')
plt.ylabel('children')
plt.show()
```

```python
# horizontal bar chart
plt.barh(colors,children,color='orange')
plt.show()
```

```python
# color and label
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'])
```

```python
# multiple bar plots & xticks
x = np.arange(len(df))
plt.bar(x - 0.2,df['2015'], width = 0.2,label='2015')
plt.bar(x ,df['2016'],width = 0.2,label='2016')
plt.bar(x + 0.2,df['2017'],width = 0.2,label='2017')
plt.xticks(x, df['batsman'])
plt.legend()
plt.show()
```

```python
# xticks --> rotation
children = [10,8,6,12,7]
colors = ['red red red red red','blue blue blue blue blue','green green green green green','yellow yellow yellow yellow','pink pink pink pink']
plt.bar(colors,children)
plt.xticks(rotation = 90)
plt.show()
```

```python
# stacked bar chat
# plt.figure(figsize=(6,4)) for size
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2015']+df['2016'],label='2017')
plt.legend()
plt.xlabel("Batsman")
plt.ylabel("Runs")
plt.xticks(rotation=45)
plt.title("Stacked Bar Chart of Runs (2015-2017)")
plt.show()
```

## Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','pst','comp']
plt.pie(data,labels=subjects)
plt.show()
```

```python
# dataset
df = pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['Red','Blue','Green','Yellow','Pink','orange'])
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0,0,0,0,0.1],shadow=True)
plt.show()
```
