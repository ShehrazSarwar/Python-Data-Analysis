###`Q 4` Using the 'Taxis' dataset - sns.load_dataset('taxis')
4.1. Create a categorical estimate plot of the totl fare - 'total' for each payment type - 'payment'.

4.2. Create a regression plot on time of ride  vs total fare. You will need to calculate ride time using pickup and dropoff column.

4.3 Give hue on payment type. and Another plot hue on taxi 'color'. Observe the plot.

```python
# code here
df1 = sns.load_dataset('taxis')
df1.head()
```

```python
# 4.1
sns.catplot(data=df1, x='payment', y='total', kind='bar',
            estimator='mean', hue='payment', legend=False)
plt.title('Payment Vs Total')
plt.show()
```

```python
# 4.2
df1['ride_time'] = (df1['dropoff'] - df1['pickup']).dt.total_seconds()/60
df1['ride_time']
```

```python
sns.regplot(data=df1, x='ride_time', y='total',
            scatter_kws={'color':'grey'})
plt.title("Regression Plot: Ride Time vs Total Fare")
plt.xlabel("Ride Time (minutes)")
plt.ylabel("Total Fare ($)")
plt.show()
```

```python
df1[['ride_time','total']].corr()
```

Regression Analysis: Ride Time vs. Total Fare

The regression plot illustrates the relationship between ride time (in minutes) and total fare (in dollars). The scatter points represent individual rides, while the fitted regression line captures the overall trend. The analysis indicates a strong positive correlation with a Pearson correlation coefficient of r = 0.8188.

This value, being close to 1, suggests that ride time is a strong predictor of total fare. In practical terms, as the duration of a ride increases, the fare charged also tends to rise proportionally. The regression line further confirms this linear association, although some variability is evident due to outliers and natural fluctuations in fare pricing.

Overall, the results demonstrate that ride time has a statistically significant and strong positive impact on total fare, aligning with expectations in real-world ride pricing models.

```python
# 4.3
sns.lmplot(data=df1, x='ride_time', y='total',hue='payment')
plt.title("Ride Time vs Total Fare by Payment Type")
plt.xlabel("Ride Time (minutes)")
plt.ylabel("Total Fare ($)")
plt.show()
```

```python
sns.lmplot(data=df1, x='ride_time', y='total', hue='color',palette='viridis')
plt.title("Ride Time vs Total Fare by Taxi Color")
plt.xlabel("Ride Time (minutes)")
plt.ylabel("Total Fare ($)")
plt.show()
```

### **`Problem 5:`** Draw a strip plot and swarm plot between "gender" and "bloodpressure" w.r.t "smoker" (use as hue parameter). Also add title to the charts.

```python
# Importing dataset
df2 = pd.read_csv('/content/insurance_data - insurance_data.csv')
df2.head()
```

```python
# strip plot
plt.figure(figsize=(15,8))
sns.stripplot(data=df2, x='gender', y='bloodpressure', hue='smoker')
plt.title('Stripplot between gender and bloodpressure')
plt.show()
```

```python
# swarmplot
plt.figure(figsize=(15,8))
sns.swarmplot(data=df2, x='gender', y='bloodpressure', hue='smoker')
plt.title('Swarmplot between gender and bloodpressure')
plt.show()
```

### **`Problem 6:`** Draw a Box-plot and a Violin plot of which x-axis represents the "region" and the y-axis represents the "bmi". Also add extra information of the column "diabetic".

```python
# Boxplot
plt.figure(figsize=(10,7))
sns.boxplot(data=df2, x='region', y='bmi', hue='diabetic')
plt.title('Boxplot between Region & Bmi (With Hue On Diabetic)')
plt.show()
```

```python
# violinplot
plt.figure(figsize=(10,7))
sns.violinplot(data=df2, x='region', y='bmi',
               hue='diabetic',
               palette='tab10',
               )
plt.title('Violinplot between Region & Bmi (With Hue On Diabetic)')
plt.show()
```

### **`Problem  7:`** Draw a bar plot and point plot of which x-axis represents the "gender" and y-axis represents "claim". Also add extra information about "smoker" column.

```python
# barplot
fig, ax = plt.subplots(1,2,sharey=True,figsize=(10,5))
sns.barplot(data=df2, x='gender', y='claim',
            hue='smoker',
            estimator='mean',
            errorbar=None, ax=ax[0])
ax[0].set_title('Barplot - Gender & Claim')
# pointplot
sns.pointplot(data=df2, x='gender', y='claim',
            hue='smoker',
            estimator='mean',
            errorbar=None, ax=ax[1])
ax[1].set_title('Pointplot - Gender & Claim')
plt.show()
```

### **`Problem 8:`** Draw a reg plot between "age" and "bmi" columns.

```python
# Regplot
sns.regplot(data=df2, x='age', y='bmi', line_kws={'color':'black'})
plt.title('Reg plot between "age" and "bmi"')
plt.show()
```

### **`Problem 10:`** Draw a pair grid of the insurance data and use "diabetic" column as a hue parameter. Also, make the diagonal plots as box-plot, upper parts as scatter plot and the lower parts as kde plot.

```python
temp_df = df2.select_dtypes(include='number')
temp_df = pd.concat([temp_df,df2['diabetic']],axis=1)
temp_df = temp_df.drop(columns=['index'])
```

```python
# code here
g = sns.PairGrid(data=temp_df, hue='diabetic', palette='viridis')
g.map_diag(sns.boxplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.add_legend()
plt.show()
```

### **`Prolem 11:`** Draw a joint plot as scatter between "bloodpressure" and "bmi". Use "smoker" as hue parameter.
