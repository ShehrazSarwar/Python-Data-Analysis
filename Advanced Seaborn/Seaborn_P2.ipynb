```python
import seaborn as sns
# import matplotlib.pyplot as plt
```

```python
# import datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

## Categorical Plots

### Categorical Scatter Plot

- Stripplot
- Swarmplot

### Categorical Distribution Plots

- Boxplot

### Categorical Estimate Plot -> for central tendency

- Barplot

### Figure level function -> `catplot`

```python
# strip plot
# axes level function
sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
# using catplot
# figure level function
sns.catplot(data=tips, x='day', y='total_bill', kind='strip')
```

```python
# jitter & hue in strip
sns.catplot(data=tips, x='day', y='total_bill', kind='strip',
            jitter=False,
            hue='sex')
```

```python
# swarmplot
sns.swarmplot(data=tips, x='day', y='total_bill')
```

```python
# using catplot
# figure level function
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm')
```

```python
# hue
sns.swarmplot(data=tips, x='day', y='total_bill', hue= 'sex')
```

```python
# In every figure level we can do facet plotting
sns.catplot(data=tips, x='day', y='total_bill',
            kind='strip',
            col='sex',
            row='time',
            hue='day')
```

### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

image.png

```python
# Box plot
# if you pass hue same as x value, it will add unique colors automatically
# you can then pass palette if you wanted to change colors
sns.boxplot(data=tips, x='day', y='total_bill',hue='day')
```

```python
# Using catplot
sns.catplot(data=tips, x='day', y='total_bill', kind='box', hue='day')
```

```python
# hue with different column
sns.catplot(data=tips, x='day', y='total_bill', kind='box', hue='sex')
```

```python
# single boxplot -> numerical col
sns.boxplot(data=tips, y='total_bill')
```

### Violinplot = (Boxplot + KDEplot)

```python
# violinplot
sns.violinplot(data=tips, x='day', y='total_bill',hue='day')
```

```python
sns.catplot(data=tips, x='day', y='total_bill',hue='day',kind='violin')
```

```python
# hue
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex')
```

```python
# hue with split=True
sns.catplot(data=tips,x='day',y='total_bill',
            kind='violin',
            hue='sex',split=True)
```

```python
# barplot
# default estimator = 'mean'
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator='min')
```

```python
# figure level
sns.catplot(data=tips, x='sex', y='total_bill', hue='smoker', kind='bar')
```

When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# to hide errorbar set errorbar=None
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', errorbar=None)
```

```python
# pointplot
sns.pointplot(data=tips, x='sex', y='total_bill')
```

```python
sns.pointplot(data=tips, x='sex', y='total_bill',hue='smoker',errorbar=None)
```

```python
# figure level
sns.catplot(data=tips, x='sex', y='total_bill',hue='smoker',kind='point')
```

A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
# countplot
sns.countplot(data=tips, x='sex',hue='smoker')
```

```python
# figure level & facet same like before
sns.catplot(data=tips, x='sex', hue='day', kind='count',
            col='smoker')
```

### Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
# line_kws & scatter_kws
sns.regplot(data=tips, x='total_bill', y='tip',
            line_kws={'color':'red'},
            scatter_kws={'color':'grey'})
```

```python
# figure level
sns.lmplot(data=tips, x='total_bill', y='tip',hue='sex')
```

```python
# residplot
sns.residplot(data=tips, x='total_bill', y='tip')
```

### A second way to plot Facet plots -> FacetGrid

```python
g = sns.FacetGrid(tips, col="time",  row="sex", hue='smoker')
g.map_dataframe(sns.scatterplot, "total_bill", "tip")
g.add_legend()
```

### Plotting Pairwise Relationship (PairGrid Vs Pairplot)

```python
# Pairplot
sns.pairplot(iris,hue='species')
```

```python
# PairGrid
g = sns.PairGrid(data=iris,hue='species')
g.map(sns.scatterplot)
```

```python
# map_diag & map_offdiag
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.kdeplot)
g.add_legend()
```

```python
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# vars
g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
g.add_legend()
```
