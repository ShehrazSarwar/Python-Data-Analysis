```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('/content/Survey Result.csv')
df.head()
```

```python
mood_df = df['Three words that decribe your mood today'].value_counts().reset_index()
```

```python
mood_df
```

```python
sns.barplot(data=mood_df, x='Three words that decribe your mood today',y='count',hue='count',palette="viridis")
plt.show()
```

```python
support_df = df['Who can you reach out to if you need support?'].fillna('None').value_counts().reset_index()
support_df
```

```python
sns.barplot(data=support_df, x='Who can you reach out to if you need support?',y='count',hue='count',palette="tab10")
plt.show()
```

```python
feel_df = df['How do you feel today? What made you feel that way?'].value_counts().reset_index()
feel_df
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=feel_df, x="count",
            y="How do you feel today? What made you feel that way?",
            hue ="How do you feel today? What made you feel that way?" ,
            palette="plasma")
plt.title("Reasons Behind Feelings", fontsize=14)
plt.xlabel("Count")
plt.ylabel("Feeling & Reason")
plt.show()
```

```python
session_df = df['How do you feel after this session?'].value_counts().reset_index()
session_df
```

```python
plt.pie(session_df['count'],labels=session_df['How do you feel after this session?'],
        autopct='%0.1f%%',explode=[0.1,0],shadow=True,startangle=90,
        colors=['#FF4A4A','#4EF5AD'])
plt.title('How do you feel after this session?')
plt.show()
```

```python
thought_df = df['Write one positive thought about yourself'].value_counts().reset_index()
thought_df
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=thought_df, x="count",
            y="Write one positive thought about yourself",
            hue ="count",
            palette="tab10",legend=True)
plt.title("One positive thought about yourself", fontsize=14)
plt.xlabel("Count")
plt.ylabel("Reason")
plt.show()
```
